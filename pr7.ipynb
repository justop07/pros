```python
def fib_search_recursive(arr, x, fib, fib1, fib2, offset):

    if fib <= 1:
        if offset + 1 < len(arr) and arr[offset + 1] == x:
            return offset + 1
        return -1

    i = min(offset + fib2, len(arr) - 1)

    if arr[i] < x:
        return fib_search_recursive(arr, x, fib1, fib2, fib1 - fib2, i)

    elif arr[i] > x:
        return fib_search_recursive(arr, x, fib2, fib1 - fib2, fib - fib1, offset)

    else:
        return i


def fibonacci_search(arr, x):
    n = len(arr)

    fib2 = 0
    fib1 = 1
    fib = fib1 + fib2

    while fib < n:
        fib2 = fib1
        fib1 = fib
        fib = fib1 + fib2

    return fib_search_recursive(arr, x, fib, fib1, fib2, -1)


# Test
arr = [10, 20, 30, 40, 50, 60, 70]
x = 70

print("Index:", fibonacci_search(arr, x))
```
