```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("CollaborativeFiltering").getOrCreate()
```

```python
df= spark.read.csv("ratings.csv", header=True, inferSchema=True)
df.show(5)
```

```python
data = df.select("userId", "itemId", "rating")
data.printSchema()
```

```python
train, test = data.randomSplit([0.8, 0.2])
```

```python
from pyspark.ml.recommendation import ALS

als = ALS(
     userCol="userId",
    itemCol="itemId",
    ratingCol="rating",
    nonnegative=True,
    implicitPrefs=False,
    coldStartStrategy="drop"
)
```

```python
model = als.fit(train)
```

```python
prediction = model.transform(test)
prediction.show(5)
```
