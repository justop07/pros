```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017/")
db = client['store_db']
products = db["products"]
```

```python
product = {
    "name": "Laptop",
    "price": 60000,
    "brand": "HP"
}
```

```python
products.insert_one(product)
```

```python
for p in products.find():
    print(p)
```

```python
for p in products.find({"name":"Laptop"}):
    print(p)
```

```python
products.delete_one({"name":"Laptop"})
```

```python
def update_product(name, new_data):
    products.update_one(
        {"name": name},
        {"$set": new_data}
    )
    print("Product updated")

# Example
update_product("iPhone 14", {"price": 75000, "stock": 8})
```
