```python
import pandas as pd

df = pd.DataFrame({
    'name': ['Soham', 'Amit'],
    'email': ['soham@gmail.com', 'amit@gmail.com'],
    'phone': ['9876543210', '9123456780']
})

# Mask email
df['email'] = df['email'].apply(lambda x: x[0] + "*****" + x[-10:])

# Mask phone
df['phone'] = df['phone'].apply(lambda x: "XXXXXX" + x[-4:])

print(df)
```

```python
# try but may work


from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("DataMasking").getOrCreate()

data = [
    ("Soham", "soham@gmail.com", "9876543210"),
    ("Amit", "amit@gmail.com", "9123456780")
]

columns = ["name", "email", "phone"]

df = spark.createDataFrame(data, columns)
df.show()

from pyspark.sql.functions import col, concat, lit, substring

df = df.withColumn(
    "phone_masked",
    concat(lit("XXXXXX"), substring(col("phone"), -4, 4))
)

df = df.withColumn(
    "email_masked",
    concat(substring(col("email"), 1, 1), lit("*****"), substring(col("email"), -10, 10))
)

df = df.withColumn(
    "name_masked",
    concat(substring(col("name"), 1, 1), lit("***"))
)

df.select("name_masked", "email_masked", "phone_masked").show()
```
