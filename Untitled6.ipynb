```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:/Users/nivyana boruah/Downloads/country_wise_latest.csv")
df
```

```python
top_cases = df.sort_values(by='Confirmed', ascending=False).head(10)
top_cases.head()
```

```python
plt.figure(figsize=(18,8))
plt.bar(top_cases['Country/Region'], top_cases['Confirmed'])
plt.xticks(rotation=45)
plt.title("Top 10 Countries by Confirmed Cases")
plt.xlabel("Country")
plt.ylabel("Confirmed Cases")
plt.show()
```

```python
top_deaths= df.sort_values(by='Deaths / 100 Cases', ascending=False).head(10)
top_deaths
```

```python
plt.figure(figsize=(10,5))
plt.bar(top_deaths['Country/Region'], top_deaths['Deaths / 100 Cases'])
plt.xticks(rotation=45)
plt.title("Highest Death Rate by Country")
plt.ylabel("Deaths per 100 Cases")
plt.show()
```
