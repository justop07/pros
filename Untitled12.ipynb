```python
data=[
    {"name":"Nivyana Boruah", "email":"ninivyan@gmail.com", "phone":"1929347580", "age":23},
    {"name": "Rahul Sharma", "email": "rahul123@yahoo.com", "phone": "9123456780", "age": 35}
]
```

```python
#Masking - email
def mask_email(email):
    name,domain = email.split("@")
    return name[0]+"****@"+ domain
```

```python
# Masking phone
def mask_phone(phone):
    return "******"+phone[-4:]
```

```python
def generalize_age(age):
    if age <18:
        return "Under 18"
    elif age< 30:
        return "18-30"
    elif age<50:
        return "30-49"
    else:
        return "50+"
```

```python
def anonymize_data(data):
    anonymize=[]
    for user in data:
        new_user={
            "name":user["name"][0]+"****",
            "email": mask_email(user["email"]),
            "phone": mask_phone(user["phone"]),
            "age": generalize_age(user["age"])
    }

        anonymize.append(new_user)
    return anonymize
```

```python
result = anonymize_data(data)
for r in result:
    print(r)
```
