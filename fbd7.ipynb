```python
# ----------------------------------
# 1. Create Spark Session
# ----------------------------------
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, avg

spark = SparkSession.builder \
    .appName("HousingAnalysis") \
    .master("local[*]") \
    .getOrCreate()

# ----------------------------------
# 2. Load Dataset
# ----------------------------------
df = spark.read.csv("Housing.csv", header=True, inferSchema=True)

# ----------------------------------
# 3. Basic Info
# ----------------------------------
print("Dataset Preview:")
df.show()

print("Schema:")
df.printSchema()

# ----------------------------------
# 4. Data Analysis
# ----------------------------------

# Average price
print("Average Price:")
df.select(avg("price")).show()

# High price houses
print("High Price Houses:")
df.filter(col("price") > 5000000).show()

# Average price by bedrooms
print("Price by Bedrooms:")
df.groupBy("bedrooms").avg("price").show()

# Furnishing impact
print("Price by Furnishing Status:")
df.groupBy("furnishingstatus").avg("price").show()

# ----------------------------------
# 5. Feature Engineering
# ----------------------------------

df = df.withColumn("price_per_sqft", col("price") / col("area"))

print("Price per Sqft:")
df.select("price_per_sqft").show()

# ----------------------------------
# 6. Stop Spark
# ----------------------------------
spark.stop()
```
