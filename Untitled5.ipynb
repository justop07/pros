```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("C:/Users/nivyana boruah/Downloads/retail_sales_dataset.csv")
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['Date']= pd.to_datetime(df['Date'])
df.head()
```

```python
df['Year']= df['Date'].dt.year
df['Month']= df['Date'].dt.month
df['Month_name']= df['Date'].dt.strftime('%b')
df['Day']= df['Date'].dt.day
```

```python
df.head()
```

```python
top_categories = df.groupby('Product Category')['Quantity'].sum()
top_categories
```

```python
import matplotlib.pyplot as plt

top_categories.plot(kind='bar')
plt.title("Top Product Categories by Quantity")
plt.xlabel("Category")
plt.ylabel("Quantity Sold")
plt.show()
```

```python
top_revenue = df.groupby('Product Category')['Total Amount'].sum()
top_revenue
```

```python
top_revenue.plot(kind="bar")
plt.title("Revenue by Category")
plt.ylabel("Total Revenue")
plt.show()
```

```python
import seaborn as sns

sns.barplot(x="Month_name", y="Total Amount", data=df, estimator=sum)
plt.title("Sales by Month")
plt.show()
```

```python
pivot = df.pivot_table(values='Total Amount', index='Month', columns='Year',aggfunc='sum')
sns.heatmap(pivot, annot=True,fmt=".0f")
plt.show()
```
